```python
import math
import numpy as np
import torch
import matplotlib.pyplot as plt
import embedded_lab as lab
from embedded_lab import table

lab.start()
```

```python
import platform
model = lab.load()
x = lab.inputs()
with torch.inference_mode():
    y = model(x)
assert x.shape == (1, 3, 48, 48) and y.shape == (1, 8) and torch.isfinite(y).all()
table(['Python', 'PyTorch', '장치', '자료형', '입력 크기', '출력 크기'],
      [[platform.python_version(), torch.__version__, 'CPU', 'FP32', tuple(x.shape), tuple(y.shape)]])
print('준비 완료')
```
