## 2. 측정
제공 함수는 03과 같은 방법으로 연산량과 시간을 측정하고 결과를 저장합니다.

```python
DEVICE_LABEL = '내 노트북 CPU'
CONDITIONS = '전원·냉각·다른 프로그램 상태를 적으세요'
settings = [('기준', 48, 1), ('입력 변경', 96, 1), ('배치 변경', 48, 8)]
trials = []
for name, size, batch in settings:
    print('측정:', name)
    trials.append(lab.run_experiment(size, batch, DEVICE_LABEL + ' / ' + name, CONDITIONS))
table(['조건', '배치 operations', '호출시간 ms', '처리량 장/s', '실효 GFLOPS'],
      [[name, f"{r['ops_batch']:,}", round(r['measurement']['median_s'] * 1000, 3),
        round(r['rates']['images_s'], 2), round(r['rates']['gflops_fp32'], 4)]
       for (name, _, _), r in zip(settings, trials)])
```

**확인:** 입력 48→96에서 연산량과 호출시간은 각각 몇 배가 되었나요?

**답:** 배치는 4배차이낫는데 호출시간은 4배까진 아님... gpus는 병렬연산에 특화가 되어있고 더 빠르게 연산하고 동시에 처리하다보니까 그런 차이가 있다. 동시에 연산하는게 많으면 시간이 막 그렇게 늘어나지 않는다고.
